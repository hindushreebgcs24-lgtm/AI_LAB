```python

from itertools import product

# Knowledge Base:
# S1: P <-> Q
# S2: Q OR R
# S3: NOT P -> R

def biconditional(P, Q):
    return P == Q

def logical_or(Q, R):
    return Q or R

def implication(A, B):
    return (not A) or B

def evaluate_kb(P, Q, R):
    S1 = biconditional(P, Q)
    S2 = logical_or(Q, R)
    S3 = implication(not P, R)

    KB = S1 and S2 and S3

    return S1, S2, S3, KB


def model_checking():
    models = []

    # Initially, assume all queries are entailed
    entail_Q = True
    entail_P_implies_R = True
    entail_R_implies_Q = True

    print("TRUTH TABLE")
    print("-" * 75)
    print(
        f"{'P':^5}{'Q':^5}{'R':^5}"
        f"{'P<->Q':^10}{'Q OR R':^10}"
        f"{'NOT P -> R':^15}{'KB':^8}"
    )
    print("-" * 75)

    # Generate all 8 interpretations of P, Q, R
    for P, Q, R in product([True, False], repeat=3):

        S1, S2, S3, KB = evaluate_kb(P, Q, R)

        def tf(value):
            return "T" if value else "F"

        print(
            f"{tf(P):^5}{tf(Q):^5}{tf(R):^5}"
            f"{tf(S1):^10}{tf(S2):^10}"
            f"{tf(S3):^15}{tf(KB):^8}"
        )

        # Select interpretations that satisfy the entire KB
        if KB:
            models.append((P, Q, R))

            # Query 1: Q
            if not Q:
                entail_Q = False

            # Query 2: P -> R
            if not implication(P, R):
                entail_P_implies_R = False

            # Query 3: R -> Q
            if not implication(R, Q):
                entail_R_implies_Q = False

    # Display the models
    print("\nMODELS OF KB")
    print("-" * 30)

    for model in models:
        print(tuple(tf(value) for value in model))

    print("\nNUMBER OF MODELS:", len(models))

    # Display entailment results
    print("\nENTAILMENT RESULTS")
    print("-" * 40)

    print("Does KB entail Q?",
          "YES" if entail_Q else "NO")

    print("Does KB entail P -> R?",
          "YES" if entail_P_implies_R else "NO")

    print("Does KB entail R -> Q?",
          "YES" if entail_R_implies_Q else "NO")


# Run the program
model_checking()
```
